# 1. Bubble Sort

## O que este código faz
O **Bubble Sort** é um algoritmo de ordenação que compara elementos adjacentes e realiza trocas quando eles estão fora de ordem.

## Objetivo desta parte
Implementar o Bubble Sort com:
- contagem de comparações;
- contagem de trocas;
- parada antecipada quando o vetor já estiver ordenado.

## Complexidade
- Melhor caso: **O(n)**
- Caso médio: **O(n²)**
- Pior caso: **O(n²)**


````python
def bubble_sort(arr):
    """Ordena uma cópia de arr usando Bubble Sort.

    Retorna:
        lista_ordenada, comparacoes, trocas
    """
    a = arr[:]
    n = len(a)

    comparacoes = 0
    trocas = 0

    for i in range(n - 1):
        trocou = False

        for j in range(n - 1 - i):
            comparacoes += 1

            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                trocas += 1
                trocou = True

        if not trocou:
            break

    return a, comparacoes, trocas
````

### Testando o Bubble Sort
A célula abaixo usa um vetor pequeno para conferir se a ordenação está funcionando como esperado.


````python
vetor_teste = [5, 1, 4, 2, 8]

ordenado, comparacoes, trocas = bubble_sort(vetor_teste)

print("Vetor original :", vetor_teste)
print("Vetor ordenado :", ordenado)
print("Comparações    :", comparacoes)
print("Trocas         :", trocas)
````

# 2. Quick Sort

## O que este código faz
O **Quick Sort** utiliza a estratégia de **dividir para conquistar**.

O algoritmo escolhe um pivô, reorganiza os elementos em torno dele e repete o processo recursivamente.

## Objetivo desta parte
Implementar o Quick Sort com:
- recursão;
- particionamento;
- contagem de comparações;
- contagem de movimentações.

## Complexidade
- Melhor caso: **O(n log n)**
- Caso médio: **O(n log n)**
- Pior caso: **O(n²)**


````python
def quick_sort(arr):
    """Ordena uma cópia de arr usando Quick Sort.

    Retorna:
        lista_ordenada, comparacoes, movimentacoes
    """
    a = arr[:]

    comparacoes = 0
    movimentacoes = 0

    def partition(lo, hi):
        nonlocal comparacoes, movimentacoes

        pivot = a[hi]
        i = lo - 1

        for j in range(lo, hi):
            comparacoes += 1

            if a[j] <= pivot:
                i += 1
                a[i], a[j] = a[j], a[i]
                movimentacoes += 1

        a[i + 1], a[hi] = a[hi], a[i + 1]
        movimentacoes += 1

        return i + 1

    def qs(lo, hi):
        if lo < hi:
            p = partition(lo, hi)
            qs(lo, p - 1)
            qs(p + 1, hi)

    qs(0, len(a) - 1)

    return a, comparacoes, movimentacoes
````

### Testando o Quick Sort


````python
vetor_teste = [5, 1, 4, 2, 8]

ordenado, comparacoes, movimentacoes = quick_sort(vetor_teste)

print("Vetor original :", vetor_teste)
print("Vetor ordenado :", ordenado)
print("Comparações    :", comparacoes)
print("Movimentações  :", movimentacoes)
````

# 3. Comparação experimental — Bubble Sort × Quick Sort

## Sobre o experimento
Os dois algoritmos são executados sobre **os mesmos vetores**, permitindo comparar a quantidade real de operações.

## Objetivo desta parte
Observar como a quantidade de operações cresce quando o tamanho da entrada aumenta.

Serão testados vetores com:
- 10 elementos;
- 20 elementos;
- 1.000 elementos.

Foi usada a semente `42` para que os mesmos dados aleatórios possam ser gerados novamente em outra execução.


````python
import random

random.seed(42)

resultados = []

for n in [10, 20, 1000]:
    original = [random.randint(0, 100000) for _ in range(n)]

    _, b_cmp, b_swap = bubble_sort(original)
    _, q_cmp, q_mov = quick_sort(original)

    resultados.append({
        "n": n,
        "bubble_comparacoes": b_cmp,
        "bubble_trocas": b_swap,
        "quick_comparacoes": q_cmp,
        "quick_movimentacoes": q_mov
    })

for r in resultados:
    print(f"\nTamanho do vetor: {r['n']}")
    print(f"Bubble Sort -> comparações: {r['bubble_comparacoes']:,} | trocas: {r['bubble_trocas']:,}")
    print(f"Quick Sort  -> comparações: {r['quick_comparacoes']:,} | movimentações: {r['quick_movimentacoes']:,}")
````

### O que os resultados mostram
O Bubble Sort apresenta crescimento **quadrático O(n²)**, enquanto o Quick Sort apresenta crescimento médio **O(n log n)**.

Por isso, conforme o tamanho do vetor aumenta, a diferença entre os dois algoritmos se torna muito mais evidente.

# 4. Busca sequencial em matriz

## O que este código faz
A busca sequencial percorre uma matriz posição por posição.

Como a matriz possui linhas e colunas, são utilizados **dois loops aninhados**.

## Objetivo desta parte
Localizar um valor e informar:
- se ele foi encontrado;
- linha;
- coluna;
- quantidade de comparações.

## Complexidade
No pior caso: **O(m × n)**


````python
def busca_sequencial_matriz(matriz, valor):
    """Procura valor em uma matriz por varredura sequencial.

    Retorna:
        encontrado, linha, coluna, comparacoes
    """
    comparacoes = 0

    linhas = len(matriz)
    colunas = len(matriz[0])

    for i in range(linhas):
        for j in range(colunas):
            comparacoes += 1

            if matriz[i][j] == valor:
                return True, i, j, comparacoes

    return False, -1, -1, comparacoes
````

### Testando a busca


````python
matriz = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

print("Busca por 10 :", busca_sequencial_matriz(matriz, 10))
print("Busca por 90 :", busca_sequencial_matriz(matriz, 90))
print("Busca por 999:", busca_sequencial_matriz(matriz, 999))
````

# 5. Experimento — Busca em matrizes de diferentes tamanhos

## Objetivo desta parte
Comparar três situações:
1. elemento na primeira posição;
2. elemento na última posição;
3. elemento inexistente.

Com isso, fica mais fácil perceber a diferença entre uma busca que termina logo no início e outra que precisa percorrer toda a matriz.


````python
def criar_matriz_sequencial(linhas, colunas):
    return [
        [i * colunas + j for j in range(colunas)]
        for i in range(linhas)
    ]
````

````python
for tamanho in [2, 10, 100]:
    matriz = criar_matriz_sequencial(tamanho, tamanho)

    primeiro = matriz[0][0]
    ultimo = matriz[-1][-1]
    inexistente = -1

    inicio = busca_sequencial_matriz(matriz, primeiro)
    fim = busca_sequencial_matriz(matriz, ultimo)
    ausente = busca_sequencial_matriz(matriz, inexistente)

    print(f"\nMatriz {tamanho} x {tamanho}")
    print("-" * 30)
    print("Total de elementos :", tamanho * tamanho)
    print("Busca no início    :", inicio[3], "comparações")
    print("Busca no fim       :", fim[3], "comparações")
    print("Valor inexistente  :", ausente[3], "comparações")
````

# 6. Hands On 1 — Array de temperaturas

## O que este código faz
O exercício trabalha com um vetor contendo 10 temperaturas.

## Objetivo desta parte
Calcular:
- média;
- maior temperatura;
- menor temperatura;
- índice do maior valor;
- índice do menor valor;
- quantidade de temperaturas acima da média.

## Complexidade
Os percursos são sequenciais, portanto a complexidade permanece **O(n)**.


````python
temperaturas = [
    19.0, 24.3, 21.3, 25.3, 25.6,
    16.1, 15.2, 29.2, 19.4, 19.0
]
````

````python
# Média
soma = 0.0

for temperatura in temperaturas:
    soma += temperatura

media = soma / len(temperaturas)

# Maior e menor valor
maior = menor = temperaturas[0]
idx_maior = idx_menor = 0

for i in range(1, len(temperaturas)):
    if temperaturas[i] > maior:
        maior = temperaturas[i]
        idx_maior = i

    if temperaturas[i] < menor:
        menor = temperaturas[i]
        idx_menor = i

# Quantidade acima da média
acima_media = 0

for temperatura in temperaturas:
    if temperatura > media:
        acima_media += 1
````

````python
print("RESULTADOS — ARRAY DE TEMPERATURAS")
print("-" * 40)
print(f"Média                : {media:.2f} °C")
print(f"Maior temperatura    : {maior:.1f} °C")
print(f"Índice do maior valor: {idx_maior}")
print(f"Menor temperatura    : {menor:.1f} °C")
print(f"Índice do menor valor: {idx_menor}")
print(f"Valores acima da média: {acima_media}")
print("Complexidade         : O(n)")
````

Uma central de distribuição recebe diariamente diversos pedidos que precisam ser organizados antes de serem encaminhados para separação e expedição. Cada pedido possui um código numérico de prioridade, e o sistema deve ordenar esses códigos do menor para o maior.

## Etapa 2 — Contagem das operações

### Insertion Sort
O **Insertion Sort** constrói o vetor ordenado da esquerda para a direita, pegando um elemento de cada vez (a *chave*) e deslocando os elementos maiores que ela uma posição à frente, até achar o lugar correto.

**Critério de contagem adotado:**
- **Comparação**: cada vez que `a[j] > chave` é avaliado dentro do laço `while`.
- **Movimentação**: cada deslocamento de um elemento uma posição à frente (`a[j+1] = a[j]`). Não contamos a inserção final da chave como movimentação, pois ela ocorre uma única vez por elemento e não representa uma comparação/deslocamento de dados já ordenados.

### Selection Sort
O **Selection Sort** percorre o vetor buscando, a cada passada, o menor elemento restante e o posiciona no início da parte ainda não ordenada.

**Critério de contagem adotado:**
- **Comparação**: cada vez que `a[j] < a[idx_min]` é avaliado ao procurar o menor elemento da parte restante.
- **Troca**: uma troca (`swap`) por passada externa, e apenas quando o menor elemento encontrado não está já na posição correta (`idx_min != i`). Esse é o motivo pelo qual o Selection Sort tende a realizar bem menos trocas que o Bubble Sort ou o Insertion Sort.

### Resumo dos critérios usados nos quatro algoritmos

| Algoritmo | O que conta como **comparação** | O que conta como **troca/movimentação** |
|---|---|---|
| Bubble Sort | `a[j] > a[j+1]` | troca de elementos adjacentes (`swap`) |
| Insertion Sort | `a[j] > chave` | deslocamento de um elemento uma posição à frente |
| Selection Sort | `a[j] < a[idx_min]` | troca do elemento da posição `i` pelo menor encontrado |
| Quick Sort | `a[j] <= pivot` no particionamento | troca de elementos durante o particionamento (inclui a troca final do pivô) |

````python
def insertion_sort(arr):
    """Ordena uma cópia de arr usando Insertion Sort.

    Retorna:
        lista_ordenada, comparacoes, movimentacoes
    """
    a = arr[:]
    n = len(a)

    comparacoes = 0
    movimentacoes = 0

    for i in range(1, n):
        chave = a[i]
        j = i - 1

        while j >= 0:
            comparacoes += 1

            if a[j] > chave:
                a[j + 1] = a[j]
                movimentacoes += 1
                j -= 1
            else:
                break

        a[j + 1] = chave

    return a, comparacoes, movimentacoes
````

````python
def selection_sort(arr):
    """Ordena uma cópia de arr usando Selection Sort.

    Retorna:
        lista_ordenada, comparacoes, trocas
    """
    a = arr[:]
    n = len(a)

    comparacoes = 0
    trocas = 0

    for i in range(n - 1):
        idx_min = i

        for j in range(i + 1, n):
            comparacoes += 1

            if a[j] < a[idx_min]:
                idx_min = j

        if idx_min != i:
            a[i], a[idx_min] = a[idx_min], a[i]
            trocas += 1

    return a, comparacoes, trocas
````

### Testando o Insertion Sort e o Selection Sort

````python
vetor_teste = [5, 1, 4, 2, 8]

ordenado, comparacoes, movimentacoes = insertion_sort(vetor_teste)
print("Insertion Sort")
print("Vetor original :", vetor_teste)
print("Vetor ordenado :", ordenado)
print("Comparações    :", comparacoes)
print("Movimentações  :", movimentacoes)

print()

ordenado, comparacoes, trocas = selection_sort(vetor_teste)
print("Selection Sort")
print("Vetor original :", vetor_teste)
print("Vetor ordenado :", ordenado)
print("Comparações    :", comparacoes)
print("Trocas         :", trocas)
````

# 10. Desafio adicional (0,5) — Efeito da organização inicial dos dados

## Objetivo
Repetir o experimento utilizando três organizações diferentes de um mesmo conjunto de dados (aqui, com n = 100 elementos):
1. um vetor com números **aleatórios**;
2. o mesmo conjunto de números, mas **já ordenado**;
3. o mesmo conjunto de números, em **ordem inversa**.

A ideia é isolar o efeito da *organização inicial dos dados* sobre a quantidade de operações, mantendo os mesmos valores e o mesmo tamanho de vetor nos três casos.

````python
random.seed(7)

n_desafio = 100
vetor_aleatorio = [random.randint(0, 100000) for _ in range(n_desafio)]
vetor_ordenado = sorted(vetor_aleatorio)
vetor_invertido = sorted(vetor_aleatorio, reverse=True)

cenarios = {
    "Aleatório": vetor_aleatorio,
    "Ordenado": vetor_ordenado,
    "Ordem inversa": vetor_invertido,
}

linhas_desafio = []

for nome_cenario, vetor_base in cenarios.items():
    _, b_cmp, b_swap = bubble_sort(vetor_base)
    _, i_cmp, i_mov  = insertion_sort(vetor_base)
    _, s_cmp, s_swap = selection_sort(vetor_base)
    _, q_cmp, q_mov  = quick_sort(vetor_base)

    linhas_desafio.append({
        "Cenário": nome_cenario,
        "Bubble Comparações": b_cmp,
        "Bubble Trocas": b_swap,
        "Insertion Comparações": i_cmp,
        "Insertion Mov.": i_mov,
        "Selection Comparações": s_cmp,
        "Selection Trocas": s_swap,
        "Quick Comparações": q_cmp,
        "Quick Mov.": q_mov,
    })

tabela_desafio = pd.DataFrame(linhas_desafio)
tabela_desafio
````

### Análise: a organização inicial dos dados interfere na quantidade de operações da mesma maneira para todos os algoritmos?

**Não, cada algoritmo reage de forma bem diferente à organização inicial dos dados:**

- **Bubble Sort** e **Insertion Sort** são os mais sensíveis à ordem inicial. No vetor já ordenado, ambos fizeram apenas 99 comparações e **zero** trocas/movimentações — o Bubble Sort para logo na primeira passada (graças à otimização de parada antecipada) e o Insertion Sort nunca precisa deslocar elementos, pois cada novo item já está na posição correta. Já no vetor em ordem inversa, esse é o **pior caso** para os dois: cada elemento precisa ser comparado e trocado/deslocado o máximo de vezes possível (4.950 comparações e praticamente o mesmo número de trocas/movimentações).

- **Selection Sort** é o mais indiferente à organização dos dados no quesito **comparações**: ele sempre faz 4.950 comparações (n(n−1)/2) para achar o mínimo a cada passada, esteja o vetor ordenado, invertido ou aleatório — o algoritmo não tem como saber que o vetor já está ordenado sem varrê-lo. Porém, o número de **trocas** ainda varia: 0 trocas no vetor já ordenado (nunca precisa mover nada), 97 no vetor aleatório e 50 no vetor invertido.

- **Quick Sort**, com a estratégia de pivô fixo (último elemento) usada aqui, é o caso mais contraintuitivo: ele se saiu **pior** exatamente no vetor já ordenado e no invertido (4.950 comparações — seu pior caso O(n²)), e melhor no vetor aleatório (apenas 651 comparações). Isso acontece porque, com vetores já ordenados (crescente ou decrescente) e pivô sempre no último elemento, as partições ficam extremamente desbalanceadas (um lado praticamente vazio e o outro com quase todos os elementos), degenerando o algoritmo para O(n²) em vez do O(n log n) esperado no caso médio.

**Conclusão do desafio:** a organização inicial dos dados **não** afeta todos os algoritmos da mesma forma. Bubble e Insertion Sort se beneficiam bastante de dados já ordenados (melhor caso O(n)) e sofrem no pior caso (ordem inversa). Selection Sort é praticamente imune à ordem dos dados em termos de comparações, mudando apenas o número de trocas. Já o Quick Sort — que na teoria é o mais rápido em média — pode se tornar o pior de todos justamente quando os dados já estão ordenados ou totalmente invertidos, caso a escolha do pivô não seja adequada. Isso mostra, na prática, a importância de estratégias melhores de escolha de pivô (como mediana de três) quando se trabalha com dados que podem vir parcial ou totalmente ordenados, como costuma acontecer em sistemas reais de pedidos.

O experimento confirmou, com dados concretos, o que a teoria de complexidade de algoritmos prevê:

- Bubble Sort, Insertion Sort e Selection Sort pertencem à mesma classe O(n²), mas **não são equivalentes na prática** — o número real de comparações e trocas/movimentações varia bastante entre eles, e o Insertion Sort costuma ser o mais eficiente dos três em dados parcialmente ordenados.
- O Quick Sort se destacou claramente para o vetor grande (n = 1.000), exigindo muito menos operações que os demais, o que o torna a escolha natural para a central de distribuição ao lidar com milhares de pedidos.

Na prática, para o sistema da central de distribuição, a recomendação é usar o **Quick Sort** (ou uma variação com melhor estratégia de escolha de pivô, para evitar o pior caso em dados já ordenados) como algoritmo padrão, reservando Bubble/Insertion/Selection Sort apenas para volumes muito pequenos de pedidos ou fins didáticos.
